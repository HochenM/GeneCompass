```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from certifi import where
```

```python
df = pd.read_csv('../data/raw/AMAT_IMPUTED_NUMERIC_3117ind_15456snp_0.05MAF_0.70coverage_0.10het.csv')

#Convert index to a column
df = df.reset_index()

df = df.rename(columns={'index':'Genotype'})
print("Dataset shape:", df.shape)
df.head()
```

```python
df
```

```python
print(df.columns[-5:])
```

```python
df_target = pd.read_csv("../data/raw/2021-2022_Across_Years_Phenomic_Data_VIS-NIR.csv"
                        )
df_target = df_target[['Genotype','DON']]
print("Dataset shape:", df_target.shape)
df_target.head()
```

```python
merge_df = pd.merge(df, df_target, left_on="Genotype", right_on="Genotype",how='inner')
```

```python
merge_df
```

```python
merge_df.isna().sum()
np.where(merge_df.isna())
```

```python
merge_df.to_csv('../data/processed/final_genomics_train_data.csv', index=False)
print("Dataset saved successfully!")
```
